```python
import matplotlib.pyplot as plt
import seaborn as sbr
```

```python
nanopore_quality_scores = seq_summary_df['mean_qscore_template']
nanopore_quality_scores_num = pd.to_numeric(nanopore_quality_scores, errors='coerce')
nanopore_quality_scores
```
